```python
import glob
import os
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
files=sorted(glob.glob("20*.csv"))
```

```python
output_file="combined_cleaned_2001_2026.csv"
```

```python
print(f"Total files found: {len(files)}")
print("Processing started...\n")
```

```python
for i,file in enumerate(files):
    temp_df=pd.read_csv(file)
    file_name = os.path.basename(file)
    year_val=os.path.splitext(file)[0]
    temp_df.insert(0,'Year',year_val)

    if i==0:
        temp_df.to_csv(output_file,mode='w',index=False)
    else:
        temp_df.to_csv(output_file,mode='a',header=False,index=False)
        
    print(f"[{i+1}/{len(files)}]Combined Year {year_val}")
print("\n Step 1 complete: saari 26 files 'combined_cleaned_2001_2026.csv' me combine ho gyi hain!")  
```

```python
input_file = "combined_cleaned_2001_2026.csv"
output_file="final_cleaned_2001_2026.csv"
```

```python
df=pd.read_csv("2021.csv")
df.head()
```

```python
chunk_size = 100000
print("Processing started in chunks...\n")
```

```python
for i, chunk in enumerate(pd.read_csv(input_file,chunksize=chunk_size)):
    num_cols = chunk.select_dtypes(include=['float64','int64']).columns
    num_cols=[c for c in num_cols if c != 'Year']
    chunk[num_cols]=chunk[num_cols].fillna(chunk[num_cols].mean())

    cat_cols = chunk.select_dtypes(include=['object']).columns
    for col in cat_cols:
        if not chunk[col].mode().empty:
            chunk[col]=chunk[col].fillna(chunk[col].mode()[0])


    for col in cat_cols:
        chunk[col]=chunk[col].astype(str).str.strip()


    if i == 0:
         chunk.to_csv(output_file,mode='w',index=False)
    else:
        chunk.to_csv(output_file,mode='a',header=False,index=False)

    print(f"Chunk {i+1} processed and appended successfully!")  

print("Entire dataset is cleaned")
```

```python
cat_cols = ['State','District','Market','Commodity','Variety','Grade']
encoders = {col: LabelEncoder() for col in cat_cols}
unique_cats={col: set() for col in cat_cols}
print("Reading text values to build mapping...")

for chunk in pd.read_csv(input_file,usecols=cat_cols,chunksize=chunk_size):
    for col in cat_cols:
        unique_cats[col].update(chunk[col].astype(str).unique())

for col in cat_cols:
    encoders[col].fit(list(unique_cats[col]))
print("Done! All text mappings are ready.")
    
```

```python
output_file='encoded_commodity_data.csv'
first_chunk=True
chunk_size=100000
for chunk in pd.read_csv(input_file,chunksize=chunk_size):
    for col in cat_cols:
        s = chunk[col].astype(str)
        mapping = dict(zip(encoders[col].classes_, range(len(encoders[col].classes_))))
        chunk[col]=s.map(mapping).fillna(-1).astype(int)
    chunk.to_csv(output_file,mode='a', index=False,header=first_chunk)
    first_chunk = False
print("Data encoding and saving complete!")    
```

```python
input_file='encoded_commodity_data.csv'
cleaned_outlier_file='outlier_handled_data.csv'

chunk_size=100000
first_chunk=True
for chunk in pd.read_csv(input_file, chunksize=chunk_size):
    num_cols=chunk.select_dtypes(include=['float64','int64']).columns

    for col in num_cols:
        if col!='Year':
            Q1 = chunk[col].quantile(0.25)
            Q3 = chunk[col].quantile(0.75)
            IQR = Q3-Q1
            lower_bound = Q1-1.5 *IQR
            upper_bound = Q3+1.5 *IQR

            chunk[col]=chunk[col].clip(lower=lower_bound,upper=upper_bound)
    chunk.to_csv(cleaned_outlier_file,mode='a',index=False,header=first_chunk)
    first_chunk=False
    
print("Outliers handling complete!")    
```

```python
from sklearn.preprocessing import StandardScaler

input_file='outlier_handled_data.csv'
scaled_output_file='scaled_commodity_data.csv'
chunk_size = 100000
first_chunk=True

scaler = StandardScaler()
for chunk in pd.read_csv(input_file,chunksize=chunk_size):
    num_cols=chunk.select_dtypes(include=['float','int64']).columns

    scale_cols = [col for col in num_cols if col !='Year']

    if len(scale_cols)>0:
        chunk[scale_cols]=scaler.fit_transform(chunk[scale_cols])
    chunk.to_csv(scaled_output_file, mode='a',index=False, header=first_chunk)    
    first_chunk=False
        
print("Feature scaling complete! Data is fully ready for model training.")


```
